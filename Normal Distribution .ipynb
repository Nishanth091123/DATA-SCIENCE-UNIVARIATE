```python
import seaborn as sns
from matplotlib import pyplot as plt
from scipy.stats import norm

def get_pdf_probability(dataset, startrange, endrange):
    # Replaced deprecated distplot with histplot
    ax = sns.histplot(dataset, kde=True, color='green', line_kws={'color': 'blue'})
    plt.axvline(startrange, color='red', linestyle='--')
    plt.axvline(endrange, color='red', linestyle='--')
    
    # Calculate sample statistics
    sample_mean = dataset.mean()
    sample_std = dataset.std()
    print('Mean=%.3f, Standard Deviation=%.3f' % (sample_mean, sample_std))
    
    # Define theoretical normal distribution
    dist = norm(sample_mean, sample_std)
    
    # Exact area between startrange and endrange using CDF
    prob = dist.cdf(endrange) - dist.cdf(startrange)
    print("The area between range({}, {}): {:.4f}".format(startrange, endrange, prob))
    
    plt.show()
    return prob

# Example call with a non-zero interval:
get_pdf_probability(dataset["ssc_p"], 40, 40)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.distributions.empirical_distribution import ECDF

# Replace with your actual path
df = pd.read_csv("PrePlacement.csv")

ecdf = ECDF(df['ssc_p'])
plt.plot(ecdf.x, ecdf.y)
plt.xlabel("ssc_p")
plt.ylabel("Cumulative Probability")
plt.title("Empirical Cumulative Distribution Function")
plt.show()
```

```python
def stdNBgraph(dataset):
        # Coverted to standard Normal Distribution
    import seaborn as sns
    mean=dataset.mean()
    std=dataset.std()

    values=[i for i in dataset]

    z_score=[((j-mean)/std) for j in values]

    sns.distplot(z_score,kde=True)

    sum(z_score)/len(z_score)
    #z_score.std()
```

```python
stdNBgraph(dataset["ssc_p"])
```
