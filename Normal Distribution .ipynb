```python
import seaborn as sns
from matplotlib import pyplot as plt
from scipy.stats import norm

def get_pdf_probability(dataset, startrange, endrange):
    # Replaced deprecated distplot with histplot
    ax = sns.histplot(dataset, kde=True, color='green', line_kws={'color': 'blue'})
    plt.axvline(startrange, color='red', linestyle='--')
    plt.axvline(endrange, color='red', linestyle='--')
    
    # Calculate sample statistics
    sample_mean = dataset.mean()
    sample_std = dataset.std()
    print('Mean=%.3f, Standard Deviation=%.3f' % (sample_mean, sample_std))
    
    # Define theoretical normal distribution
    dist = norm(sample_mean, sample_std)
    
    # Exact area between startrange and endrange using CDF
    prob = dist.cdf(endrange) - dist.cdf(startrange)
    print("The area between range({}, {}): {:.4f}".format(startrange, endrange, prob))
    
    plt.show()
    return prob

# Example call with a non-zero interval:
get_pdf_probability(dataset["ssc_p"], 40, 40)
```
