```python
def freqTable(columnName, dataset):
    table = pd.DataFrame(
        columns=["Unique_Values", "Frequency", "Relative Frequency", "Cusum"]
    )

    table["Unique_Values"] = dataset[columnName].value_counts().index
    table["Frequency"] = dataset[columnName].value_counts().values
    table["Relative Frequency"] = table["Frequency"] / table["Frequency"].sum()
    table["Cusum"] = table["Relative Frequency"].cumsum()

    return table
```

```python
def Univariate(dataset, quan):

    table = pd.DataFrame(
        index=[
            "Mean",
            "Median",
            "Mode",
            "Q1:25%",
            "Q2:50%",
            "Q3:75%",
            "99%",
            "Q4:100%",
            "IQR",
            "1.5rule",
            "Lesser",
            "Greater",
            "Min",
            "Max"
        ],
        columns=quan
    )

    for column in quan:

        mean = dataset[column].mean()
        median = dataset[column].median()
        mode = dataset[column].mode()[0]

        q1 = dataset[column].quantile(0.25)
        q2 = dataset[column].quantile(0.50)
        q3 = dataset[column].quantile(0.75)
        q99 = dataset[column].quantile(0.99)
        q4 = dataset[column].quantile(1)

        iqr = q3 - q1
        rule = 1.5 * iqr

        lesser = q1 - rule
        greater = q3 + rule

        minimum = dataset[column].min()
        maximum = dataset[column].max()

        table.loc["Mean", column] = mean
        table.loc["Median", column] = median
        table.loc["Mode", column] = mode
        table.loc["Q1:25%", column] = q1
        table.loc["Q2:50%", column] = q2
        table.loc["Q3:75%", column] = q3
        table.loc["99%", column] = q99
        table.loc["Q4:100%", column] = q4
        table.loc["IQR", column] = iqr
        table.loc["1.5rule", column] = rule
        table.loc["Lesser", column] = lesser
        table.loc["Greater", column] = greater
        table.loc["Min", column] = minimum
        table.loc["Max", column] = maximum

    return table
```

```python
import numpy as np
import pandas as pd

quan = ["sl_no", "ssc_p", "hsc_p", "degree_p", "etest_p", "mba_p", "salary"]


def Univariate(dataset, quan):
  descriptive = pd.DataFrame(
      index=[
          "Mean",
          "Median",
          "Mode",
          "Q1:25%",
          "Q2:50%",
          "Q3:75%",
          "99%",
          "Q4:100%",
          "IQR",
          "1.5rule",
          "Lesser",
          "Greater",
          "Min",
          "Max",
          "kurtosis",
          "skew",
      ],
      columns=quan,
  )

  for columnName in quan:
    descriptive.loc["Mean", columnName] = dataset[columnName].mean()
    descriptive.loc["Median", columnName] = dataset[columnName].median()
    descriptive.loc["Mode", columnName] = dataset[columnName].mode()[0]
    descriptive.loc["Q1:25%", columnName] = dataset.describe()[columnName][
        "25%"
    ]
    descriptive.loc["Q2:50%", columnName] = dataset.describe()[columnName][
        "50%"
    ]
    descriptive.loc["Q3:75%", columnName] = dataset.describe()[columnName][
        "75%"
    ]
    descriptive.loc["99%", columnName] = np.percentile(
        dataset[columnName], 99
    )  # produces NaN for salary if missing values exist
    descriptive.loc["Q4:100%", columnName] = dataset.describe()[columnName][
        "max"
    ]
    descriptive.loc["IQR", columnName] = (
        descriptive.loc["Q3:75%", columnName]
        - descriptive.loc["Q1:25%", columnName]
    )
    descriptive.loc["1.5rule", columnName] = (
        1.5 * descriptive.loc["IQR", columnName]
    )
    descriptive.loc["Lesser", columnName] = (
        descriptive.loc["Q1:25%", columnName]
        - descriptive.loc["1.5rule", columnName]
    )
    descriptive.loc["Greater", columnName] = (
        descriptive.loc["Q3:75%", columnName]
        + descriptive.loc["1.5rule", columnName]
    )
    descriptive.loc["Min", columnName] = dataset[columnName].min()
    descriptive.loc["Max", columnName] = dataset[columnName].max()
    descriptive.loc["kurtosis", columnName] = dataset[columnName].kurtosis()
    descriptive.loc["skew", columnName] = dataset[columnName].skew()

  return descriptive


result = Univariate(dataset, quan)
result
```
